# ✍️ AI 에이전트 & Claude Code 특강 — 느낀 점 & 적용 방법

> 강의 내용 요약은 [01_강의요약.ipynb](./01_강의요약.ipynb)에 있습니다.

## 1. 느낀 점

### KPT

**① 강의 정리 방식**
- **K**: 특강을 녹음해 두고 전사본을 AI로 정리하는 방식은 계속 유지하려고 합니다. 녹음이 없었다면 보고서 작성과 복습이 어려웠을 것입니다. 확실하지 않은 내용은 웹 자료로 2차 검증했습니다.
- **P**: AI가 정리한 결과에 제가 실제로 하지 않은 실습 기록이 섞여 있었습니다. 터미널에서 Claude CLI를 실행했다는 내용이 그 예입니다. 이를 통해 AI 정리에도 **환각(hallucination)이 발생할 수 있다**는 것을 깨달았습니다. 전사본 자체에도 강사의 말실수나 전사 오류가 섞여 있었습니다. 예를 들어 "코인베이스"는 실제로는 Hugging Face 해킹 사건이었습니다.
- **T**: 매 수업의 정리 노트는 노션 MCP와 Claude로 정리하되, 결과물은 **제가 직접 검토해 환각을 방지**하겠습니다. 특히 실습 기록처럼 제 경험이 들어가는 부분은 실제로 한 일인지 하나씩 확인하겠습니다. 이 노트가 그 첫 번째 결과물입니다.

**② 로그인·환경 세팅**
- **K**: 문제가 생겼을 때 AI와 함께 해결하는 방식이 효과적이었습니다. 집에서 노션 연결 오류가 있었으나 AI와 협업하며 해결했습니다.
- **P**: 로그인 과정에서 어려움을 겪었습니다. 계정 인증 코드 입력 단계와 노션 인증 시 403 오류가 발생했고, 안티그래비티 설치와 Claude 로그인에 예상보다 많은 시간이 걸렸습니다.
- **T**: 수업 전에 미리 할 수 있는 세팅은 사전에 마쳐 두겠습니다. 새로 세팅할 때는 Claude 전용 폴더에서 시작하고, 로그인은 시크릿 창에서 진행해 오류를 줄이겠습니다.

**③ 계정·권한 보안**
- **K**: 연결하기 전에 "이 연결이 누구에게까지 열리는가"를 확인했습니다. 덕분에 계정이 공유되고 있다는 점을 알아차리고, 이 PC에서만 쓰는 MCP 방식으로 바꿀 수 있었습니다. 앞으로도 이 습관을 유지하겠습니다.
- **P**: 토큰과 권한 개념을 제대로 모른 채 설정하다 보니, 토큰을 채팅창에 붙여 넣는 실수를 했고 권한 구조를 이해하지 못해 여러 번 헤맸습니다. (자세한 과정은 실습 3 참조)
- **T**: 새 도구를 연결할 때는 권한이 어떤 구조인지 먼저 이해하고, **필요한 최소 권한만** 주겠습니다. 강의에서 배운 MCP 보안 주의사항을 실제로 적용하는 것입니다.

**④ 도구 숙련도**
- **K**: 안티그래비티로 코딩해 본 경험 덕분에 기본 사용법은 익숙했고, MCP 활용법과 Skill 생성법은 처음 접하는 내용이라 유익했습니다.
- **P**: Claude Code와 노션 사용법이 아직 익숙하지 않습니다. 그동안 API 호출 위주로만 코딩해 와서, 새 파일과 폴더를 만들어 로컬 환경을 처음부터 구성하는 작업이 낯설었습니다.
- **T**: Claude Code로 로컬 폴더에서 직접 작업해 보며 익숙해지겠습니다. 또 수업에서 소개된 슬라이드 생성 Skill을 시작으로 여러 Skill을 익혀, 공부와 프로젝트에 바로 활용해 보겠습니다.

### AAR (After Action Review)

- **① 원래 목표는 무엇이었는가**: 널리 알려진 Claude Code의 사용법을 배우는 것이 목표였습니다.
- **② 실제 결과는 어떠했는가**: 이전에 안티그래비티로 코딩을 해 본 경험이 있어 수업 자체는 비교적 쉽게 느껴졌습니다. 다만 MCP 활용법과 Skill 생성법은 처음 접하는 내용이어서 유익했습니다. 배운 내용을 바로 적용하여 본 정리 노트를 노션 MCP로 작성했습니다.
- **③ 차이가 발생한 이유는 무엇인가**: 기본 사용법은 안티그래비티 경험 덕분에 익숙했으나, 로그인·환경 세팅·파일 기반 작업은 처음이라 막히는 부분이 있었습니다. API 호출 위주로만 코딩해 온 습관 때문에 로컬 환경 구성이 낯설었던 것이 원인이라고 판단합니다.
- **④ 다음에는 무엇을 다르게 할 것인가**: 전용 폴더에서 처음부터 시작하고, 시크릿 창에서 로그인하며, 설치와 로그인은 수업 전에 미리 마쳐 두겠습니다. 강의 내용은 AI로 1차 정리하더라도, 제가 직접 읽어 보며 환각이 없는지 검토하겠습니다. 또한 이번 경험을 통해 목표가 명확하다면 AI와 협업하여 해결하지 못할 문제는 없다는 점을 느꼈으므로, 막히더라도 우선 시도해 보는 태도를 유지하겠습니다.

## 2. 적용 방법

### 2-1. 강의 내용을 직접 적용한 실습

#### 실습 1. 안티그래비티 & Claude 로그인

| 단계 | 내용 | 결과 |
|---|---|---|
| 1 | 안티그래비티(Antigravity) 설치 | ✅ |
| 2 | Claude.ai 구독 계정으로 로그인 | ✅ |
| 3 | 안티그래비티에서 Claude를 활용해 실습 진행 | ✅ |

**막혔던 부분 & 해결**

- 안티그래비티 설치와 Claude 로그인에 예상보다 시간이 많이 걸림 → 다음부터는 수업 전에 미리 세팅해 두기로 함
- 로그인 과정의 **계정 인증 코드 입력** 단계에서 시간이 걸림 → 이후에는 시크릿 창에서 로그인하기로 함

#### 실습 2. Notion MCP 연결 (집에서 연결 → MCP로 재연결)

| 단계 | 내용 | 결과 |
|---|---|---|
| 1 | 집에서 Claude에 Notion 연결 → 브라우저에서 노션 로그인, 워크스페이스 접근 허용 | ✅ |
| 2 | 강의 녹음 전사본(txt 3개)을 주고 정리 요청 → 노션 페이지 자동 생성 | ✅ 이 문서의 원본 |
| 3 | 계정이 **공유되고 있다는 점**을 확인 → 계정 단위 연결을 해제하기로 함 | ✅ |
| 4 | 전용 폴더에 `.mcp.json`(`https://mcp.notion.com/mcp`)을 만들어 **이 PC에서만** 쓰는 MCP로 다시 연결 | ✅ |
| 5 | 재연결 후 Claude가 노션 페이지를 정상적으로 읽는지 확인 | ✅ |

**막혔던 부분 & 해결**

- 노션 인증 중 `403` 오류 → 재시도로 해결
- 처음 연결한 **claude.ai 계정 커넥터**는 계정 단위라서, 같은 Claude 계정으로 로그인한 **다른 사람·다른 기기에서도 내 노션에 접근 가능** → 계정이 공유되고 있어 위험하다고 판단. 계정 연결을 해제하고, 인증 토큰이 **이 PC에만 저장되는** `.mcp.json` 방식으로 전환
- 연결 직후에는 이미 열려 있던 세션에 노션 도구가 바로 보이지 않음 → 새 세션을 열어 해결

**강의에서 배운 것을 그대로 적용한 지점**

- 강사가 시연한 "파일 첨부 + 노션 링크 → 정리해줘" 워크플로우([강의 요약 §9](./01_강의요약.ipynb#9-실전-사례--notion-mcp))를 그대로 재현
- MCP 보안 주의([강의 요약 §11](./01_강의요약.ipynb#11-️-mcp-보안-주의사항)): Notion 공식 MCP 서버(`mcp.notion.com`)만 사용

#### 실습 3. GitHub MCP 연결 & 정리 노트 업로드

Notion을 MCP로 다시 연결한 김에, 같은 `.mcp.json`에 GitHub도 MCP로 추가했습니다. 그리고 노션에 정리한 이 특강 노트를 이 저장소에 올렸습니다.

| 단계 | 내용 | 결과 |
|---|---|---|
| 1 | `.mcp.json`에 GitHub 공식 MCP 서버(`https://api.githubcopilot.com/mcp/`) 추가, 인증 헤더는 `Bearer ${GITHUB_PAT}` 환경변수 참조 | ✅ |
| 2 | GitHub Fine-grained PAT 발급 → Windows 사용자 환경변수 `GITHUB_PAT`에 저장 | ✅ |
| 3 | Claude가 노션 페이지를 MCP로 읽어 Markdown으로 변환 → `강의요약` / `느낀점_적용방법` 2개 파일로 분리 | ✅ |
| 4 | 저장소 `cv-special-lecture` 생성 → 같은 토큰으로 `git push` | ✅ 이 저장소 |

**막혔던 부분 & 해결**

- 토큰을 **채팅창에 그대로 붙여넣는 실수** → 대화 기록에 남아 노출된 것으로 보고 즉시 폐기 후 재발급. 이후에는 환경변수 편집 창에서만 입력
- `.mcp.json`에 토큰을 직접 쓰지 않고 `${GITHUB_PAT}`로 참조 → 폴더가 OneDrive로 동기화돼도 토큰은 노출되지 않음
- 환경변수 저장 후에도 도구가 안 보임 → 앱이 켜진 상태에서 저장했기 때문. 앱을 **완전히 재시작**해야 반영됨
- 저장소 생성 `403 Resource not accessible by personal access token` → 토큰에 저장소 생성 권한이 없음. 빈 저장소는 GitHub 웹에서 직접 생성
- `git push` `403` → 토큰의 Repository access가 **"Public repositories"(읽기 전용)** 로 되어 있었음. **Only select repositories → `cv-special-lecture`** 선택 + **Contents: Read and write** 권한 추가로 해결
  - 여기서 여러 번 헤맨 이유: **저장소를 선택하지 않은 채** 권한만 추가하려고 했음. Repository access가 Public repositories로 남아 있으면 **Repositories 권한 탭 자체가 나타나지 않아서**, Add permissions에서 **Account 권한 목록만** 보였고 Contents를 찾을 수 없었음
  - 그 과정에서 푸시와 무관한 **Account 권한 21개**를 잘못 추가했다가 모두 삭제. 토큰을 새로 만들었을 때도 기본값(Public repositories)을 바꾸지 않아 같은 문제가 반복됨
  - 결론: **① 저장소 선택(Only select repositories) → ② Repositories 탭에서 권한 추가 → ③ Update** 순서를 지켜야 함. 권한이 실제로 저장됐는지는 토큰 상세 페이지에서 다시 확인

**강의에서 배운 것을 그대로 적용한 지점**

- MCP 보안 주의([강의 요약 §11](./01_강의요약.ipynb#11-️-mcp-보안-주의사항))를 실제 설정에 반영: 공식 서버만 사용, 토큰은 **필요한 저장소 1개 + 필요한 권한 1개(Contents)** 로 최소화, 만료 기간 설정
- "All repositories"도 고려했지만, 에이전트가 늘 쓰는 토큰이라 잘못 동작하거나 유출될 경우 피해 범위를 줄이기 위해 **저장소 1개만** 허용

### 2-2. 향후 계획 및 소감

- 별도의 프로그램을 만들기보다는, 수업에서 소개된 슬라이드 생성 Skill을 비롯해 여러 가지 Skill을 익혀 공부와 프로젝트의 능률을 높이는 것이 목표입니다.
- 진로에 대한 고민은 여전히 남아 있습니다. 교수님의 조언에 따라 우선 여러 가지를 직접 시도해 보면서 방향을 찾아가고자 합니다.
